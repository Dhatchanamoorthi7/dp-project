### K-Means Clustering on Market Basket Data

First, let's load the `Market_Basket_Optimisation.csv` dataset and inspect its structure. Market basket data often represents transactions, where each row lists items purchased in a single transaction. For K-Means, we usually need a numerical representation, often a binary matrix indicating the presence or absence of items.

```python
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer

# Load the dataset
# Assuming the CSV has one transaction per row, with items separated by commas or similar delimiter
# We'll read it in a way that treats each item as a separate entry.
# For market basket data, it's common for each row to be a list of items bought in a transaction.
# Let's assume the CSV is structured such that each line is a transaction with items separated by commas.

# Read the data, ensuring no header and handling potential missing values as empty strings
with open('/content/Market_Basket_Optimisation.csv', 'r') as f:
    transactions = f.readlines()

# Clean up and split each transaction into a list of items
transactions = [t.strip().split(',') for t in transactions]

# Flatten the list of transactions to get all unique items
all_items = [item.strip() for transaction in transactions for item in transaction if item.strip() != '']
unique_items = sorted(list(set(all_items)))

print(f"Number of transactions: {len(transactions)}")
print(f"Number of unique items: {len(unique_items)}")

# Display first few transactions
print("\nFirst 5 transactions:")
for i, t in enumerate(transactions[:5]):
    print(f"Transaction {i+1}: {t}")
```

```python
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_items = pd.DataFrame(te_ary, columns=te.columns_)

print("\nOne-hot encoded DataFrame head:")
display(df_items.head())

# Convert boolean DataFrame to integer (0s and 1s) for K-Means
df_items_numeric = df_items.astype(int)

print("\nNumeric (0/1) DataFrame head:")
display(df_items_numeric.head())
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# Determine the optimal number of clusters using the Elbow Method
max_k = 10 # Let's test up to 10 clusters
inertia = []

for i in range(1, max_k + 1):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10) # n_init is important for robust results
    kmeans.fit(df_items_numeric)
    inertia.append(kmeans.inertia_)

# Plot the elbow curve
fig = plt.figure(figsize=(10, 6))
plt.plot(range(1, max_k + 1), inertia, marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.grid(True)
plt.xticks(range(1, max_k + 1))
plt.show()
```

```python
from sklearn.decomposition import PCA

k = 4 # Example: Choose k based on the elbow method. Adjust as needed.

kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
clusters = kmeans.fit_predict(df_items_numeric)

# Add cluster labels to the original (or one-hot encoded) DataFrame
df_items_numeric['Cluster'] = clusters

print(f"Distribution of transactions per cluster (for k={k}):")
display(df_items_numeric['Cluster'].value_counts().sort_index())

# To visualize, we often need to reduce dimensionality (e.g., using PCA)
pca = PCA(n_components=2) # Reduce to 2 dimensions for plotting
principal_components = pca.fit_transform(df_items_numeric.drop('Cluster', axis=1))

pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
pca_df['Cluster'] = clusters

fig = plt.figure(figsize=(12, 8))
sns.scatterplot(
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='viridis',
    data=pca_df,
    legend='full',
    alpha=0.6
)
plt.title(f'2D PCA of K-Means Clusters (K={k})')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```

```python
print(f"\nAnalyzing characteristics of each cluster (K={k}):")

# Calculate the mean presence of each item for each cluster
cluster_means = df_items_numeric.groupby('Cluster').mean()

for i in range(k):
    print(f"\nCluster {i} (n={df_items_numeric['Cluster'].value_counts()[i]} transactions):")
    # Get top 10 items for each cluster based on mean presence
    top_items = cluster_means.loc[i].sort_values(ascending=False).head(10)
    display(top_items.rename('Mean Presence'))
```
